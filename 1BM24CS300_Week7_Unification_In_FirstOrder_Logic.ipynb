```python

def is_variable(x):
    return isinstance(x, str) and x.islower()

def apply_substitution(term, subst):

    if isinstance(term, str):
        while term in subst:
            term = subst[term]
        return term

    return tuple(apply_substitution(x, subst) for x in term)

def occurs(variable, term, subst):

    term = apply_substitution(term, subst)

    if variable == term:
        return True

    if isinstance(term, tuple):
        return any(occurs(variable, x, subst) for x in term)

    return False

def unify(x, y, subst=None):

    if subst is None:
        subst = {}

    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    if x == y:
        return subst

    if is_variable(x):

        if occurs(x, y, subst):
            return None

        subst[x] = y
        return subst

    if is_variable(y):

        if occurs(y, x, subst):
            return None

        subst[y] = x
        return subst

    if not isinstance(x, tuple) or not isinstance(y, tuple):
        return None

    if x[0] != y[0]:
        return None

    if len(x) != len(y):
        return None

    for a, b in zip(x[1:], y[1:]):

        result = unify(a, b, subst)

        if result is None:
            return None

        subst = result

    return subst


def test_unification(expr1, expr2):

    print("\nExpression 1:", expr1)
    print("Expression 2:", expr2)

    result = unify(expr1, expr2)

    if result is None:
        print("Unification Failed")

    else:
        print("Unification Successful")

        if result:
            print("Substitutions:")
            for variable, value in result.items():
                print(variable, "=", value)
        else:
            print("No substitution required (NIL)")

        print("Unified Expression:",
              apply_substitution(expr1, result))

print("\nCASE 1: Successful Unification")

expr1 = ("Eats", "x", "Apple")
expr2 = ("Eats", "Riya", "y")

test_unification(expr1, expr2)

print("\nCASE 2: Different Predicate Symbols")

expr1 = ("Eats", "x", "Apple")
expr2 = ("Likes", "Riya", "Apple")

test_unification(expr1, expr2)

print("\nCASE 3: Different Number of Arguments")

expr1 = ("Likes", "x", "Apple")
expr2 = ("Likes", "Riya")

test_unification(expr1, expr2)

print("\nCASE 4: Occurs Check Failure")

expr1 = ("P", "x")
expr2 = ("P", ("f", "x"))

test_unification(expr1, expr2)
```
