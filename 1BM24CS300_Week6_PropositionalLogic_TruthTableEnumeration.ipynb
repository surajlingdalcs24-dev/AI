```python
import pandas as pd
truth_values = [True, False]
rows = []

for P in truth_values:
    for Q in truth_values:
        for R in truth_values:

            sent1 = not Q or P
            sent2 = not P or not Q
            sent3 = Q or R

            kb_true = sent1 and sent2 and sent3

            entail_R = R
            entail_R_to_P = not R or P
            entail_Q_to_R = not Q or R
            rows.append({
                'P': P,
                'Q': Q,
                'R': R,
                'Q -> P': sent1,
                'P -> ~Q': sent2,
                'Q v R': sent3,
                'KB': kb_true,
                'Target: R': entail_R,
                'Target: R -> P': entail_R_to_P,
                'Target: Q -> R': entail_Q_to_R
            })

df = pd.DataFrame(rows)
display(df)

kb_models = df[df['KB'] == True]
print("\nModels where the KB is True (KB is Satisfied):")
display(kb_models[['P', 'Q', 'R']])

for target in ['Target: R', 'Target: R -> P', 'Target: Q -> R']:
    entails = kb_models[target].all()
    print(f"Does KB entail {target.replace('Target: ', '')}? {entails}")
```

### i) Truth Table for the Knowledge Base (KB)
This truth table shows the truth value of the individual sentences in the KB ($Q \rightarrow P$, $P \rightarrow \neg Q$, and $Q \lor R$) and highlights the models (interpretations) where the entire KB evaluates to True (satisfied).

```python
import pandas as pd

truth_values = [True, False]
rows_kb = []

for P in truth_values:
    for Q in truth_values:
        for R in truth_values:
            sent1 = not Q or P           # Q -> P
            sent2 = not P or not Q       # P -> ~Q
            sent3 = Q or R               # Q v R
            kb_true = sent1 and sent2 and sent3

            rows_kb.append({
                'P': P, 'Q': Q, 'R': R,
                'Q -> P': sent1,
                'P -> ~Q': sent2,
                'Q v R': sent3,
                'KB is True?': kb_true
            })

df_kb = pd.DataFrame(rows_kb)
display(df_kb)

print("\nModels where KB is True:")
display(df_kb[df_kb['KB is True?'] == True][['P', 'Q', 'R']])
```

### ii) Truth Table and Entailment Analysis for Target: $R$
This truth table shows the truth values of the KB alongside the target $R$ to evaluate if $KB \vDash R$ (i.e., whenever KB is True, is $R$ also True?).

```python
rows_r = []
for P in truth_values:
    for Q in truth_values:
        for R in truth_values:
            kb_true = (not Q or P) and (not P or not Q) and (Q or R)
            rows_r.append({
                'P': P, 'Q': Q, 'R': R,
                'KB (Premise)': kb_true,
                'Target: R (Conclusion)': R,
                'KB |= R?': not kb_true or R
            })

df_r = pd.DataFrame(rows_r)
display(df_r)
```

### iii) Truth Table and Entailment Analysis for Target: $R \rightarrow P$
This truth table evaluates whether $KB \vDash (R \rightarrow P)$.

```python
rows_r_to_p = []
for P in truth_values:
    for Q in truth_values:
        for R in truth_values:
            kb_true = (not Q or P) and (not P or not Q) and (Q or R)
            target = not R or P          # R -> P
            rows_r_to_p.append({
                'P': P, 'Q': Q, 'R': R,
                'KB (Premise)': kb_true,
                'Target: R -> P (Conclusion)': target,
                'KB |= (R -> P)?': not kb_true or target
            })

df_r_to_p = pd.DataFrame(rows_r_to_p)
display(df_r_to_p)
```
